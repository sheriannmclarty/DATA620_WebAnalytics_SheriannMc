For this first assignment, I'm creating a basic graph using the Krackhardt kite network. This is a classic example in social network analysis that shows different positions and roles within a small network.

The Krackhardt kite has 10 nodes and demonstrates concepts like centrality and structural holes that we'll be studying throughout the semester.

```python
# Import necessary libraries
import networkx as nx
import matplotlib.pyplot as plt

# Setting up the plot size so the graph is easier to see
plt.rcParams['figure.figsize'] = [10, 8]
```

## Building the Krackhardt Kite Network

I'm going to manually create the nodes and edges for this network. The kite network has 10 people, which I'll label with names (Diane through Jane) to make it more intuitive than just using numbers.

```python
# Initialize an empty graph
G = nx.Graph()

# Add nodes - representing 10 individuals in the network
nodes = ['Diane', 'Beverley', 'Andre', 'Carol', 'Ed',
         'Fernando', 'Garth', 'Heather', 'Ike', 'Jane']

G.add_nodes_from(nodes)

print(f"Number of nodes: {G.number_of_nodes()}")
print(f"Nodes: {list(G.nodes())}")
```

```python
# Add edges - these define the relationships in the kite network
# The pattern here creates the distinctive kite shape with a tail

edges = [
    ('Diane', 'Beverley'),
    ('Diane', 'Andre'),
    ('Diane', 'Carol'),
    ('Diane', 'Fernando'),
    ('Beverley', 'Andre'),
    ('Beverley', 'Ed'),
    ('Beverley', 'Fernando'),
    ('Andre', 'Carol'),
    ('Andre', 'Ed'),
    ('Andre', 'Fernando'),
    ('Carol', 'Ed'),
    ('Ed', 'Fernando'),
    ('Ed', 'Garth'),
    ('Fernando', 'Garth'),
    ('Fernando', 'Heather'),
    ('Garth', 'Heather'),
    ('Heather', 'Ike'),
    ('Ike', 'Jane')
]

G.add_edges_from(edges)

print(f"Number of edges: {G.number_of_edges()}")
```

## Visualizing the Network

Now let's plot the network to see the kite structure. I'm using NetworkX's spring layout which should show the structure pretty well.

```python
# Create the visualization
plt.figure(figsize=(12, 9))

# Use spring layout for positioning - it tends to produce nice results
pos = nx.spring_layout(G, seed=42)  # seed for reproducibility

# Draw the network
nx.draw(G, pos,
        with_labels=True,
        node_color='lightblue',
        node_size=1500,
        font_size=11,
        font_weight='bold',
        edge_color='gray',
        width=2)

plt.title('Krackhardt Kite Network', fontsize=16, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## Basic Network Statistics

Since we're just getting started with graph analysis, I wanted to explore a few basic properties of this network to better understand its structure.

```python
# Calculate some basic network metrics

# Degree - how many connections each person has
degrees = dict(G.degree())
print("Degree (number of connections) for each node:")
for node, degree in sorted(degrees.items(), key=lambda x: x[1], reverse=True):
    print(f"  {node}: {degree}")
```

```python
# Check if the graph is connected
print(f"\nIs the graph connected? {nx.is_connected(G)}")

# Calculate average clustering coefficient
avg_clustering = nx.average_clustering(G)
print(f"Average clustering coefficient: {avg_clustering:.3f}")

# Network density - what proportion of all possible edges actually exist
density = nx.density(G)
print(f"Network density: {density:.3f}")
```

## Observations

Looking at the degree distribution, Fernando is clearly the most central person in the network with 6 connections, followed by Andre and Ed who each have 5 connections. On the other end, Jane is only connected to one person (Ike), making her quite peripheral.

The network is connected overall, which means there's a path between any two people. But you can see from the visualization that Jane and Ike are kind of hanging off to the side - they form what's called a "tail" in the kite structure.
